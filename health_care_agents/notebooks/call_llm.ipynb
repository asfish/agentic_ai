```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

```python
llm = ChatGoogleGenerativeAI(model='gemma-4-31b-it')
response = llm.invoke('Please list all the planets of the solar system from bigest to smallest')

print(response.text)
```

```python
response = llm.invoke([
    ['system', 'you are an amazing space expert asystant that answers concisely.'], 
    ['human', 'which planet does have the most moon and how much does it have?'],
])
print(response.text)
```

```python
from google import genai

client = genai.Client()
response = client.models.generate_content(
    model='gemini-3.5-flash',
    contents='Explain how gen ai works in few words, and include how it uses transformer behind the scenes.',
)
print(response.text)
```

# LLM models from Groq
* [Quick start](https://console.groq.com/docs/quickstart)
* [Create key](https://console.groq.com/keys)
* [groq models](https://console.groq.com/docs/models)
    

```python
from langchain_groq import ChatGroq
g_llm = ChatGroq(
    model_name="qwen/qwen3-32b",
    reasoning_format="parsed",
    temperature=0.7
)

msg = [
    (
        "system",
        "You are a helpful assistant that gives a short answer to a question",
    ),
    (
        "human",
        "How many moons does Jupiter have?"
    ),
]
g_response = g_llm.invoke(msg)
print(g_response.content)
```
